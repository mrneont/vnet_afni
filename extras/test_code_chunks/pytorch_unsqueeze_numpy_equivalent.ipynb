```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import torch
```

```python
t1_orig  = nib.load('/Users/yamunasn/RR_AFNI_VNET/mridataset/data_00_basic_iso_128/training/orig/pac_125_orig.nii.gz')
t1_data  = np.asanyarray(t1_orig.dataobj).astype('float32')
```

```python
t1_torch = torch.from_numpy(t1_data)
```

```python
# numpy and torch tensor : original data
print('torch_tensor size = ',t1_torch.size())
print('numpy_arr size = ',t1_data.shape)
```

```python
# ref: https://note.nkmk.me/en/python-numpy-newaxis/
# numpy unsqueezed in 1 x 1 x 128 x 128 x 128
t1_data_unsq = np.expand_dims(t1_data, 0)
t1_data_unsq = np.expand_dims(t1_data_unsq, 0)
print('t1_data_unsq shape =', t1_data_unsq.shape)
```

```python
# torch tensor unsqueezed in 1 x 1 x 128 x 128 x 128
t1_torch_unsq = torch.unsqueeze(t1_torch,0) 
t1_torch_unsq = torch.unsqueeze(t1_torch_unsq,0) 
print('torch_tensor size = ',t1_torch_unsq.size())
```

```python
#convert t1_data_unsq(numpy array unsq) to torch tensor 
torch_from_np = torch.from_numpy(t1_data_unsq)
```

```python
#differnce between torch unsqueezed and np unsqueezed
res = t1_torch_unsq - torch_from_np
```
