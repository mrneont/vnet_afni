```python
def get_dice(pred, gt):
    # num is the total number of classes, include the background
    dice = 2*np.sum(pred*gt)/(np.sum(pred+gt))
            
    return dice
```

```python
def get_continous_dice(pred, gt,c):
    # num is the total number of classes, include the background
    #print(c)
    
    dice = 2*np.sum(pred*gt)/((c*np.sum(gt))+np.sum(pred))
            
    return dice
```

```python
def get_c(gt, pred):
    # num is the total number of classes, include the background
    #dice = np.sum(pred*gt)/(np.sum(pred)+np.sum(gt))
    c = np.sum(pred*gt)/(np.sum(gt*np.sign(pred)))
            
    return c
```
